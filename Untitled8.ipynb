```python
import pandas as pd
import numpy as np
import re
import ipaddress
import joblib
```

```python
df = pd.read_csv("dataset.zip")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
print("Total URLs:", len(df))

print("\nLabel distribution:")
print(df["label"].value_counts())

print("\nFirst 5 URLs:")
print(df["URL"].head())
```

```python
def extract_url_features(url):

    if pd.isna(url):
        url = ""

    url = str(url).strip()

    if not re.match(r"^https?://", url, re.IGNORECASE):
        url = "https://" + url

    parsed = urlparse(url)

    domain = parsed.netloc.split("@")[-1].split(":")[0]

    url_length = len(url)
    domain_length = len(domain)

    # Check whether domain is an IP address
    try:
        ipaddress.ip_address(domain)
        is_domain_ip = 1
    except ValueError:
        is_domain_ip = 0

    # Count subdomains
    domain_parts = domain.split(".")
    no_of_subdomain = max(len(domain_parts) - 2, 0)

    # Characters
    letters = sum(c.isalpha() for c in url)
    digits = sum(c.isdigit() for c in url)

    no_of_equals = url.count("=")
    no_of_qmark = url.count("?")
    no_of_ampersand = url.count("&")

    # Special characters
    special_chars = sum(
        not c.isalnum()
        and c not in [".", "/", "=", "?", "&", "-", "_"]
        for c in url
    )

    # Obfuscation
    obfuscated_chars = url.count("%")
    has_obfuscation = int(obfuscated_chars > 0)

    # Ratios
    letter_ratio = letters / url_length if url_length else 0
    digit_ratio = digits / url_length if url_length else 0
    special_ratio = special_chars / url_length if url_length else 0
    obfuscation_ratio = obfuscated_chars / url_length if url_length else 0

    # HTTPS
    is_https = int(parsed.scheme.lower() == "https")

    return {
        "URLLength": url_length,
        "DomainLength": domain_length,
        "IsDomainIP": is_domain_ip,
        "NoOfSubDomain": no_of_subdomain,
        "HasObfuscation": has_obfuscation,
        "NoOfObfuscatedChar": obfuscated_chars,
        "ObfuscationRatio": obfuscation_ratio,
        "NoOfLettersInURL": letters,
        "LetterRatioInURL": letter_ratio,
        "NoOfDegitsInURL": digits,
        "DegitRatioInURL": digit_ratio,
        "NoOfEqualsInURL": no_of_equals,
        "NoOfQMarkInURL": no_of_qmark,
        "NoOfAmpersandInURL": no_of_ampersand,
        "NoOfOtherSpecialCharsInURL": special_chars,
        "SpacialCharRatioInURL": special_ratio,
        "IsHTTPS": is_https
    }

print("Feature extractor created successfully!")
```

```python
test_url = df["URL"].iloc[0]

test_features = extract_url_features(test_url)

print("URL:")
print(test_url)

print("\nExtracted features:")
for key, value in test_features.items():
    print(key, ":", value)
```

```python
feature_rows = []

for url in df["URL"]:
    feature_rows.append(extract_url_features(url))

X_raw = pd.DataFrame(feature_rows)

y_raw = df["label"].copy()

print("Feature extraction completed!")
print("X shape:", X_raw.shape)
print("y shape:", y_raw.shape)

print("\nFeatures:")
print(X_raw.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_raw,
    y_raw,
    test_size=0.20,
    random_state=42,
    stratify=y_raw
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

raw_url_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

raw_url_model.fit(X_train, y_train)

print("Raw URL model trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

y_pred = raw_url_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
train_pred = raw_url_model.predict(X_train)
test_pred = raw_url_model.predict(X_test)

print("Training Accuracy:", accuracy_score(y_train, train_pred))
print("Testing Accuracy :", accuracy_score(y_test, test_pred))
```

```python
deploy_features = X_raw.columns.tolist()

joblib.dump(raw_url_model, "ShopSafe_Raw_URL_Model.pkl")
joblib.dump(deploy_features, "ShopSafe_Raw_URL_Features.pkl")

print("Model saved successfully!")
print("ShopSafe_Raw_URL_Model.pkl")
print("ShopSafe_Raw_URL_Features.pkl")
```

```python
def predict_raw_url(url):

    feature_dict = extract_url_features(url)

    input_df = pd.DataFrame([feature_dict])

    input_df = input_df[deploy_features]

    prediction = raw_url_model.predict(input_df)[0]
    probabilities = raw_url_model.predict_proba(input_df)[0]

    class_probabilities = dict(
        zip(raw_url_model.classes_, probabilities)
    )

    phishing_probability = class_probabilities.get(0, 0)
    legitimate_probability = class_probabilities.get(1, 0)

    if prediction == 1:
        result = "LEGITIMATE"
    else:
        result = "PHISHING"

    return {
        "URL": url,
        "Prediction": result,
        "Phishing Probability": round(phishing_probability * 100, 2),
        "Legitimate Probability": round(legitimate_probability * 100, 2)
    }


result = predict_raw_url("https://www.google.com")

print(result)
```

```python
test_urls = [
    "https://www.google.com",
    "https://www.amazon.com",
    "https://www.microsoft.com",
    "http://example.com"
]

for url in test_urls:
    print(predict_raw_url(url))
```
